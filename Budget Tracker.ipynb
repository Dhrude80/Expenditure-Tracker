```python
Total_budget = int(input("Enter your monthly budget"))
Categories = {
    1:"food",
    2:"clothes",
    3:"rent",
    4:"investing",
    5:"schemes",
    6:"entertainment",
    7:"other_essentials",
    8:"others"
}

def percent_expenditure(amount):
    return (amount*100)/Total_budget  
    
while True:
    total_expenditure = 0
    total_percent = 0
    over_budget = False
    for category in Categories:
        y = Categories[category]
        b = int(input(f"Enter your {y} expenditure"))
        x = percent_expenditure(b)
        if x > 100:
            print("Invalid")
            continue
        total_expenditure += b
        total_percent += x
        if total_expenditure > Total_budget:
            print("Total Expenditure exceeds the budget")
            over_budget = True
            break
        print(f"{y} : {x}")
    if over_budget == False:
        print(f"You total expenditure is {total_expenditure} and you have {100-total_percent} % of your income left.")
```
